```python
import json
import pandas as pd
import os

with open("kloppy/tests/files/footovision_tracking.json") as f:
    raw_frame = json.load(f)

new_dict = {}

df_players = (
    pd.DataFrame(raw_frame["players"])
    .sort_values(by=["halftime", "time"])
    .reset_index(drop=True)
)

new_raw_players = []

for halftime in df_players["halftime"].unique():
    df_half = df_players[df_players["halftime"] == halftime].copy()

    start_time = df_half["time"].min()
    small_half = df_half[df_half["time"] <= start_time + 100_000]

    new_raw_players.extend(small_half.to_dict(orient="records"))



df_ball = pd.DataFrame(raw_frame["ball"])

new_raw_ball = []

for halftime in df_ball["halftime"].unique():
    df_half = df_ball[df_ball["halftime"] == halftime].copy()

    start_time = df_half["time"].min()
    small_half = df_half[df_half["time"] <= start_time + 100_000]

    new_raw_ball.extend(small_half.to_dict(orient="records"))


new_dict["players"] = new_raw_players
new_dict["ball"] = new_raw_ball

with open(
    "kloppy/tests/files/footovision_small_tracking.json",
    "w",
) as outfile:
    json.dump(new_dict, outfile)

size_mb = os.path.getsize(
    "kloppy/tests/files/footovision_small_tracking.json"
) / 1024**2

print(f"{size_mb:.2f} MB")
```
