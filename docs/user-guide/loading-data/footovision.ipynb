# Footovision


## Event data

Footovision event data is loaded from an events feed and the corresponding match metadata. The loader returns a standardized `EventDataset`; coordinates can be left in Footovision's source system or transformed to another supported provider coordinate system.


```python
from kloppy import footovision

dataset = footovision.load_event(
    event_data="../../../kloppy/tests/files/footovision_events.json",
    meta_data="../../../kloppy/tests/files/footovision_game.json",
    coordinates="footovision",
)

dataset.events[:5]
```

```python
from kloppy import footovision

dataset = footovision.load_tracking(
    meta_data="../../../kloppy/tests/files/footovision_game.json",
    raw_data="../../../kloppy/tests/files/footovision_small_tracking.json",
    coordinates="footovision",
    limit=100,
    only_alive=False,
)

dataset.to_df().head()
```

```python
frame = dataset.frames[0]

print(dataset.metadata)
print(frame.ball_coordinates)
print(frame.players_coordinates)
```
